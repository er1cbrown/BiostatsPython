```python
import codecs

# Detect encoding and convert ALLSONGS.txt to UTF-8
try:
    with open('/content/ALLSONGS.txt', 'rb') as source_file:
        raw_data = source_file.read(4)
        # Check for UTF-16 Byte Order Marks
        if raw_data.startswith(codecs.BOM_UTF16_BE) or raw_data.startswith(codecs.BOM_UTF16_LE):
            encoding = 'utf-16'
        else:
            encoding = 'utf-16' # Fallback attempt for common Windows exports

    with open('/content/ALLSONGS.txt', 'r', encoding=encoding) as source_file:
        content = source_file.read()

    with open('/content/ALLSONGS.txt', 'w', encoding='utf-8-sig') as target_file:
        target_file.write(content)
    print("File successfully converted to UTF-8-sig.")
except Exception as e:
    print(f"Conversion failed or not needed: {e}")
```

### Step 3: Main Conversion Script
Processes the folder of MP3s and compiles them into compatible multi-channel STEM format.

```python
import subprocess
import tempfile
import numpy as np
import soundfile as sf
import stempeg
from pathlib import Path
from tqdm.auto import tqdm

def mp3_folder_to_stem_files(input_folder, output_folder, model="htdemucs_ft"):
    input_folder = Path(input_folder)
    output_folder = Path(output_folder)
    output_folder.mkdir(parents=True, exist_ok=True)

    mp3_files = sorted(input_folder.glob("*.mp3")) + sorted(input_folder.glob("*.MP3"))
    if not mp3_files:
        print("No MP3 files found.")
        return

    print(f"Found {len(mp3_files)} MP3s\n")

    for mp3 in tqdm(mp3_files, desc="Processing tracks"):
        print(f"\n→ {mp3.name}")

        with tempfile.TemporaryDirectory() as tmpdir:
            tmp = Path(tmpdir)

            # 1. Separate with Demucs (GPU if available)
            cmd = [
                "demucs",
                "-n", model,
                "--out", str(tmp),
                str(mp3),
            ]
            result = subprocess.run(cmd, capture_output=True, text=True)
            if result.returncode != 0:
                print(f"  Demucs failed: {result.stderr[-500:]}")
                continue

            stem_dir = tmp / model / mp3.stem
            if not stem_dir.exists():
                print("  Stem folder not found")
                continue

            try:
                # 2. Load original + 4 stems
                # NI STEM order: mix, drums, bass, other, vocals
                mix, sr = sf.read(mp3, always_2d=True)
                drums, _ = sf.read(stem_dir / "drums.wav", always_2d=True)
                bass, _ = sf.read(stem_dir / "bass.wav", always_2d=True)
                other, _ = sf.read(stem_dir / "other.wav", always_2d=True)
                vocals, _ = sf.read(stem_dir / "vocals.wav", always_2d=True)

                min_len = min(len(mix), len(drums), len(bass), len(other), len(vocals))
                stems = np.stack([
                    mix[:min_len],
                    drums[:min_len],
                    bass[:min_len],
                    other[:min_len],
                    vocals[:min_len],
                ], axis=0)

                # 3. Write proper NI STEM file
                out_path = output_folder / f"{mp3.stem}.stem.m4a"
                stempeg.write_stems(
                    str(out_path),
                    stems,
                    sample_rate=sr,
                    writer=stempeg.NIStemsWriter(
                        codec="aac",
                        bitrate=256000,
                    ),
                )
                print(f"  ✓ Saved → {out_path.name}")

            except Exception as e:
                print(f"  Packaging error: {e}")

    print(f"\nDone! STEM files are in:\n{output_folder}")

# Run the extraction
mp3_folder_to_stem_files(INPUT_FOLDER, OUTPUT_FOLDER, MODEL)
```
