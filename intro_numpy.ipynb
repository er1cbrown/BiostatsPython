# NumPy Essentials for Data Science

Welcome to **NumPy**! Short for *Numerical Python*, NumPy is the fundamental library for scientific computing and data science in Python.

### Why learn NumPy?
1. **Speed:** It performs mathematical operations much faster than standard Python lists.
2. **Simplicity:** It allows you to run operations across entire datasets without writing explicit `for` loops.
3. **Foundation:** Virtually all data science tools in Python (like Pandas, Matplotlib, and Scikit-Learn) are built on top of NumPy arrays.

---

## 1. Importing NumPy

To use NumPy, we first need to import it. By standard data science convention, we import `numpy` and alias it as `np` to save typing.

```python
# Step 1: Import the numpy library under the standard alias 'np'
import numpy as np

# Step 2: Verify that NumPy is installed by checking its version
print("NumPy version:", np.__version__)
```

## 2. Vectors: Creation and Conversion

In data science, a 1-dimensional collection of numbers is called a **vector** (or a 1D NumPy array). In this section, you will learn how to convert regular Python lists to NumPy vectors, convert them back, and generate new vectors automatically.

### Step 2.1: Converting a Python List into a NumPy Vector

```python
# 1. Define a standard Python list containing numbers
my_list = [10, 20, 30, 40, 50]

# 2. Convert the Python list into a NumPy vector using np.array()
my_vector = np.array(my_list)

# 3. Print both to see how they differ in format
print("Original Python List:", my_list)
print("Converted NumPy Vector:", my_vector)

# 4. Check the data type to verify the conversion
print("Type of vector object:", type(my_vector))
```

### Step 2.2: Converting a NumPy Vector back into a Python List

```python
# 1. Use the .tolist() method on the NumPy array object
restored_list = my_vector.tolist()

# 2. Inspect the result
print("Restored Python List:", restored_list)
print("Type of restored object:", type(restored_list))
```

### Step 2.3: Creating Vectors Directly with Built-in Functions

```python
# Example A: Create a range of numbers (start=1, stop=10, step=2)
range_vec = np.arange(1, 10, 2)
print("Vector with arange(1, 10, 2):", range_vec)

# Example B: Create 5 evenly spaced numbers between 0 and 1
linear_vec = np.linspace(0, 1, 5)
print("Vector with linspace(0, 1, 5):", linear_vec)

# Example C: Create a vector filled with zeros
zeros_vec = np.zeros(4)
print("Vector of zeros:", zeros_vec)

# Example D: Create a vector filled with ones
ones_vec = np.ones(4)
print("Vector of ones:", ones_vec)
```

## 3. Element-Wise Math & Broadcasting

In standard Python, adding numbers to a list requires a `for` loop or list comprehension. In NumPy, math operations are automatically applied to every element in the array simultaneously (this feature is called **broadcasting**).

```python
# Define a sample vector
scores = np.array([50, 60, 70, 80, 90])

# 1. Add a scalar value to every element
print("Add 5 bonus points:", scores + 5)

# 2. Multiply every element by 2
print("Double all scores:", scores * 2)

# 3. Perform element-wise addition between two vectors of equal length
extra_credit = np.array([2, 4, 1, 5, 3])
final_scores = scores + extra_credit
print("Total scores after adding extra credit:", final_scores)
```

## 4. Calculating Basic Summary Statistics

Data analysis often starts with understanding key summary statistics of a variable (such as average, variation, minimum, and maximum).

```python
# Sample dataset: Monthly income values (in thousands)
incomes = np.array([45, 50, 52, 48, 55, 60, 52, 120])  # 120 is a potential outlier

# Step 1: Calculate Total Sum
print("Total Sum:", np.sum(incomes))

# Step 2: Calculate Central Tendency (Mean vs Median)
print("Mean Income:", np.mean(incomes))
print("Median Income:", np.median(incomes))

# Step 3: Calculate Measures of Spread (Standard Deviation and Variance)
print("Standard Deviation:", np.std(incomes))
print("Variance:", np.var(incomes))

# Step 4: Calculate Minimum, Maximum, and Percentiles
print("Minimum Income:", np.min(incomes))
print("Maximum Income:", np.max(incomes))
print("25th Percentile:", np.percentile(incomes, 25))
print("75th Percentile:", np.percentile(incomes, 75))
```

## 5. Essential NumPy Functions for Data Science

Here are four fundamental capabilities every data science student should master early on.

### Step 5.1: Finding Locations of Min/Max Values (`argmax` / `argmin`)
Sometimes you don't just want the highest number—you need to know **which position (index)** contains the highest or lowest number.

```python
temperatures = np.array([72, 75, 68, 91, 84, 79])

# Find the index position of the highest temperature
hottest_day_index = np.argmax(temperatures)

# Find the index position of the lowest temperature
coolest_day_index = np.argmin(temperatures)

print("Hottest temperature:", temperatures[hottest_day_index], "at index:", hottest_day_index)
print("Coolest temperature:", temperatures[coolest_day_index], "at index:", coolest_day_index)
```

### Step 5.2: Filtering Data with Boolean Indexing
You can easily filter data by passing conditional expressions inside square brackets `[]`.

```python
ages = np.array([14, 21, 18, 42, 12, 33, 65])

# 1. Create a True/False boolean mask based on a condition
is_adult_mask = ages >= 18
print("Boolean Mask:", is_adult_mask)

# 2. Apply the mask to extract only the values where the condition is True
adult_ages = ages[ages >= 18]
print("Filtered Adults Array:", adult_ages)
```

### Step 5.3: Reshaping Vectors into Matrices
Data models frequently require restructuring 1D arrays into 2D matrices.

```python
# Create a 1D vector of 6 elements
flat_vector = np.arange(1, 7)  # [1, 2, 3, 4, 5, 6]
print("Original 1D Shape:", flat_vector.shape)

# Reshape the 1D vector into a 2D matrix (2 rows, 3 columns)
matrix_2d = flat_vector.reshape((2, 3))
print("Reshaped 2D Matrix:\n", matrix_2d)
print("New 2D Shape:", matrix_2d.shape)
```

### Step 5.4: Generating Random Numbers
Generating random datasets is vital for experiments, model initialization, and data sampling.

```python
# Set a random seed so that the random outputs are repeatable across runs
np.random.seed(42)

# 1. Generate 5 random integers between 10 and 100
random_integers = np.random.randint(10, 100, size=5)
print("Random Integers:", random_integers)

# 2. Generate 5 random floats from a Standard Normal Distribution (mean=0, std=1)
normal_samples = np.random.randn(5)
print("Normal Distribution Samples:", normal_samples)
```
